```python
import pandas as pd
import zipfile
import os

files = os.listdir('.')

zip_files = [f for f in files if f.endswith('.zip')]
for z in zip_files:
    try:
        with zipfile.ZipFile(z, 'r') as zip_ref:
            zip_ref.extractall('.')
        print(f"Successfully extracted: {z}")
    except Exception as e:
        print(f"Could not extract {z}: {e}")

data_files = [f for f in os.listdir('.') if f.endswith(('.csv', '.xls', '.xlsx')) and not f.startswith('sample')]
print("\nData files found:", data_files)

if data_files:
    target_file = data_files[0]
    try:
        df = pd.read_csv(target_file, encoding='utf-8')
    except (UnicodeDecodeError, Exception):
        df = pd.read_csv(target_file, encoding='windows-1252')

    df.columns = [col.strip() for col in df.columns]

    print("\n--- Column Names ---")
    print(df.columns.tolist())

    print("\n--- Dataset Sample ---")
    print(df.head(3))
else:
    print("No CSV or Excel files found. Please check your left sidebar uploads!")
```

```python
import pandas as pd
import numpy as np

time_cols = [c for c in df.columns if any(k in c.lower() for k in ['date', 'time', 'timestamp'])]
for col in time_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

session_cols = [c for c in df.columns if 'session' in c.lower() and 'id' in c.lower()]
session_col = session_cols[0] if session_cols else None

user_cols = [c for c in df.columns if 'user' in c.lower() or 'visitor' in c.lower() or 'client' in c.lower()]
user_col = user_cols[0] if user_cols else None

bounce_cols = [c for c in df.columns if 'bounce' in c.lower()]
bounce_col = bounce_cols[0] if bounce_cols else None

duration_cols = [c for c in df.columns if 'duration' in c.lower() or 'time_spent' in c.lower()]
duration_col = duration_cols[0] if duration_cols else None

total_sessions = df[session_col].nunique() if session_col else len(df)
total_users = df[user_col].nunique() if user_col else (df[session_col].nunique() if session_col else len(df))

if bounce_col:
    if df[bounce_col].dtype == 'object':
        bounce_rate = (df[bounce_col].astype(str).str.lower().isin(['true', '1', 'yes']).mean()) * 100
    else:
        bounce_rate = df[bounce_col].mean() * (100 if df[bounce_col].max() <= 1 else 1)
else:
    bounce_rate = 0.0

avg_duration = df[duration_col].mean() if duration_col else 0.0

traffic_metrics = pd.DataFrame({
    'Metric': ['Total Sessions', 'Total Unique Users', 'Bounce Rate (%)', 'Avg Session Duration (sec)'],
    'Value': [f"{total_sessions:,}", f"{total_users:,}", f"{bounce_rate:.2f}%", f"{avg_duration:.1f}s"]
})

print("=== WEB TRAFFIC CORE METRICS ===")
print(traffic_metrics.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style="whitegrid")

landing_cols = [c for c in df.columns if any(k in c.lower() for k in ['landing', 'entry', 'page', 'url'])]
landing_col = landing_cols[0] if landing_cols else df.columns[0]

source_cols = [c for c in df.columns if any(k in c.lower() for k in ['source', 'channel', 'medium', 'referrer'])]
source_col = source_cols[0] if source_cols else None

time_col = time_cols[0] if time_cols else None

if time_col:
    df['DateOnly'] = df[time_col].dt.date
    daily_traffic = df.groupby('DateOnly').size().reset_index(name='Sessions')

    plt.figure(figsize=(12, 5))
    plt.plot(daily_traffic['DateOnly'], daily_traffic['Sessions'], color='#2b5c8f', linewidth=2)
    plt.title('Website Traffic Volume Over Time', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Date', fontsize=11)
    plt.ylabel('Total Sessions', fontsize=11)
    plt.tight_layout()
    plt.savefig('traffic_trends_over_time.png', dpi=300)
    plt.show()

top_pages = df[landing_col].value_counts().head(10).reset_index()
top_pages.columns = [landing_col, 'Sessions']

plt.figure(figsize=(10, 5))
sns.barplot(data=top_pages, x='Sessions', y=landing_col, palette='Blues_r', hue=landing_col, legend=False)
plt.title('Top Landing Pages by Traffic Volume', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Sessions', fontsize=11)
plt.ylabel('Landing Page / Route', fontsize=11)
plt.tight_layout()
plt.savefig('top_landing_pages.png', dpi=300)
plt.show()

if source_col:
    top_sources = df[source_col].value_counts().head(8).reset_index()
    top_sources.columns = [source_col, 'Sessions']

    plt.figure(figsize=(10, 5))
    sns.barplot(data=top_sources, x=source_col, y='Sessions', palette='Viridis', hue=source_col, legend=False)
    plt.title('Traffic Acquisition Channels & Referral Sources', fontsize=14, fontweight='bold', pad=15)
    plt.xlabel('Referral Channel / Source', fontsize=11)
    plt.ylabel('Sessions', fontsize=11)
    plt.tight_layout()
    plt.savefig('traffic_sources.png', dpi=300)
    plt.show()
```

```python
!pip install -q reportlab

import os
from reportlab.lib.pagesizes import letter
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, Image, HRFlowable
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle

pdf_filename = 'Website_Traffic_Analysis_Report.pdf'
doc = SimpleDocTemplate(pdf_filename, pagesize=letter, rightMargin=36, leftMargin=36, topMargin=36, bottomMargin=36)
story = []

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    'DocTitle',
    parent=styles['Heading1'],
    fontName='Helvetica-Bold',
    fontSize=20,
    leading=24,
    textColor=colors.HexColor('#1A365D'),
    spaceAfter=4
)

subtitle_style = ParagraphStyle(
    'DocSubtitle',
    parent=styles['Normal'],
    fontName='Helvetica',
    fontSize=10,
    leading=12,
    textColor=colors.HexColor('#4A5568'),
    spaceAfter=12
)

h2_style = ParagraphStyle(
    'SectionHeader',
    parent=styles['Heading2'],
    fontName='Helvetica-Bold',
    fontSize=13,
    leading=16,
    textColor=colors.HexColor('#1A365D'),
    spaceBefore=10,
    spaceAfter=6
)

body_style = ParagraphStyle(
    'BodyDark',
    parent=styles['Normal'],
    fontName='Helvetica',
    fontSize=9,
    leading=12,
    textColor=colors.HexColor('#2D3748'),
    spaceAfter=6
)

story.append(Paragraph("Website Traffic & User Journey Analysis", title_style))
story.append(Paragraph("Alfido Tech Analytics | Traffic Acquisition, Page Flows & Conversion Optimization Report", subtitle_style))
story.append(HRFlowable(width="100%", thickness=1.5, color=colors.HexColor('#2B6CB0'), spaceAfter=10))

story.append(Paragraph("1. Executive Traffic Metrics Summary", h2_style))

kpi_data = [
    [Paragraph("<b>Metric</b>", body_style), Paragraph("<b>Value</b>", body_style)],
    [Paragraph("Total Sessions", body_style), Paragraph(f"{total_sessions:,}", body_style)],
    [Paragraph("Total Unique Users", body_style), Paragraph(f"{total_users:,}", body_style)],
    [Paragraph("Overall Bounce Rate", body_style), Paragraph(f"{bounce_rate:.2f}%", body_style)],
    [Paragraph("Average Session Duration", body_style), Paragraph(f"{avg_duration:.1f}s", body_style)]
]

t_kpi = Table(kpi_data, colWidths=[240, 240])
t_kpi.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (1, 0), colors.HexColor('#EBF8FF')),
    ('TEXTCOLOR', (0, 0), (1, 0), colors.HexColor('#2B6CB0')),
    ('ALIGN', (0, 0), (-1, -1), 'LEFT'),
    ('BOTTOMPADDING', (0, 0), (-1, -1), 4),
    ('TOPPADDING', (0, 0), (-1, -1), 4),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#CBD5E0')),
]))
story.append(t_kpi)
story.append(Spacer(1, 10))

story.append(Paragraph("2. Traffic Volume & Landing Page Distribution", h2_style))
if os.path.exists('traffic_trends_over_time.png'):
    story.append(Image('traffic_trends_over_time.png', width=500, height=190))
    story.append(Spacer(1, 8))

if os.path.exists('top_landing_pages.png'):
    story.append(Image('top_landing_pages.png', width=500, height=200))
    story.append(Spacer(1, 10))

story.append(Paragraph("3. Acquisition Channels & Referrals", h2_style))
if os.path.exists('traffic_sources.png'):
    story.append(Image('traffic_sources.png', width=500, height=190))
    story.append(Spacer(1, 10))

story.append(Paragraph("4. 5 Conversion Optimizations for Alfido Tech", h2_style))

opt_1 = "<b>1. Optimize High-Dropoff Landing Pages:</b> Streamline UX and reduce load speeds on top entry pages experiencing elevated bounce rates to retain initial session momentum."
opt_2 = "<b>2. Strategic Call-To-Action (CTA) Placement:</b> Position above-the-fold CTAs on top-traffic landing routes to guide users deeper into high-converting funnel pages."
opt_3 = "<b>3. Re-engage High-Volume Referral Traffic:</b> Tailor landing experiences for visitors arriving from top referral channels with personalized welcome banners or offer codes."
opt_4 = "<b>4. Exit-Intent Popups & Retargeting:</b> Deploy targeted exit-intent prompts on key conversion pages to re-capture abandoning visitors before they bounce."
opt_5 = "<b>5. Mobile & Speed Optimization:</b> Conduct cross-device friction audits to improve mobile responsiveness and ensure fast page load times across all traffic segments."

for opt in [opt_1, opt_2, opt_3, opt_4, opt_5]:
    story.append(Paragraph(f"• {opt}", body_style))
    story.append(Spacer(1, 2))

doc.build(story)
print("PDF Report generated successfully: Website_Traffic_Analysis_Report.pdf")
```
